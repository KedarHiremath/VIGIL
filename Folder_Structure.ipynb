```python
from pathlib import Path

def print_directory_tree(dir_path, prefix="", ignore_dirs=None):
    """
    Recursively prints the directory structure, ignoring specified directories.
    """
    if ignore_dirs is None:
        ignore_dirs = {".git", ".pytest_cache", ".ruff_cache", ".venv", "storage_data", "__pycache__"}

    path = Path(dir_path)
    
    try:
        # Get all contents, sort directories first, then files alphabetically
        contents = sorted(path.iterdir(), key=lambda x: (x.is_file(), x.name.lower()))
    except PermissionError:
        return

    # Filter out the directories we want to ignore
    contents = [c for c in contents if not (c.is_dir() and c.name in ignore_dirs)]
    
    # Prepare formatting pointers
    pointers = ["├── "] * (len(contents) - 1) + ["└── "] if contents else []
    
    for pointer, path_obj in zip(pointers, contents):
        print(f"{prefix}{pointer}{path_obj.name}")
        
        if path_obj.is_dir():
            # Extend the prefix for the next level
            extension = "│   " if pointer == "├── " else "    "
            print_directory_tree(path_obj, prefix + extension, ignore_dirs)

if __name__ == "__main__":
    # Point this to your VIGIL folder path, or use "." for the current directory
    root_folder = "."  
    
    print(Path(root_folder).resolve().name)
    print_directory_tree(root_folder)
```

```python
import os
from pathlib import Path

def extract_code_contents(target_dir="src", output_file="Project_Codes_with_Path.txt"):
    root_path = Path(target_dir)
    
    if not root_path.exists() or not root_path.is_dir():
        print(f"Error: Directory '{target_dir}' does not exist.")
        return

    separator = "\n" + "-" * 80 + "\n"
    
    # Open the output file in append mode ('a')
    with open(output_file, "a", encoding="utf-8") as out_f:
        # Iterate through all items recursively
        for file_path in sorted(root_path.rglob("*")):
            # Skip if it's a directory
            if not file_path.is_file():
                continue
                
            # Ignore __pycache__ directories
            if "__pycache__" in file_path.parts:
                continue
                
            # Try reading the file (handles potential non-text files gracefully)
            try:
                with open(file_path, "r", encoding="utf-8") as in_file:
                    content = in_file.read()
                    
                # Write in the exact format requested directly to the file
                out_f.write(f"PATH - {file_path}\n")
                out_f.write("CODE:\n")
                out_f.write(content)
                # Ensure there's a newline before the separator if the file doesn't end with one
                if not content.endswith('\n'):
                    out_f.write("\n")
                out_f.write(separator)
                
            except UnicodeDecodeError:
                # Skip binary files that can't be read as text
                pass
            except Exception as e:
                out_f.write(f"Error reading {file_path}: {e}\n")
                out_f.write(separator)
                
    print(f"Successfully appended contents to {output_file}")

if __name__ == "__main__":
    extract_code_contents("src")
```

```python
import os
from pathlib import Path

def extract_root_files(target_dir=".", output_file="Project_Codes_with_Path.txt"):
    root_path = Path(target_dir)
    
    # Set of files to explicitly ignore
    # Note: We also ignore the output_file so it doesn't try to append to itself
    ignore_files = {".env", "Folder_Structure.ipynb", output_file}

    separator = "\n" + "-" * 80 + "\n"
    
    # Open the output file in append mode ('a')
    with open(output_file, "a", encoding="utf-8") as out_f:
        # iterdir() only looks at the immediate directory (no subdirectories)
        for file_path in sorted(root_path.iterdir()):
            # Skip if it's a directory
            if not file_path.is_file():
                continue
                
            # Skip the explicitly ignored files
            if file_path.name in ignore_files:
                continue
                
            # Try reading the file (handles potential non-text files gracefully)
            try:
                with open(file_path, "r", encoding="utf-8") as in_file:
                    content = in_file.read()
                    
                # Write in the exact format requested directly to the file
                out_f.write(f"PATH - {file_path}\n")
                out_f.write("CODE:\n")
                out_f.write(content)
                
                # Ensure there's a newline before the separator if the file doesn't end with one
                if not content.endswith('\n'):
                    out_f.write("\n")
                out_f.write(separator)
                
            except UnicodeDecodeError:
                # Skip binary files that can't be read as text (like .lock files if they are binary)
                pass
            except Exception as e:
                out_f.write(f"Error reading {file_path}: {e}\n")
                out_f.write(separator)
                
    print(f"Successfully appended contents from root files to {output_file}")

if __name__ == "__main__":
    # Call the function for the root directory (current directory ".")
    extract_root_files(".")
```

```python
import tiktoken

def count_tokens(file_path="Project_Codes_with_Path.txt"):
    try:
        # 'o200k_base' is the tokenizer used by the latest GPT-4o models
        # You can also use 'cl100k_base' for older models
        encoding = tiktoken.get_encoding("o200k_base")
        
        with open(file_path, "r", encoding="utf-8") as f:
            content = f.read()
            
        # Encode the text into tokens and count them
        token_count = len(encoding.encode(content))
        
        print(f"File size: {len(content):,} characters")
        print(f"Total tokens: {token_count:,}")
        
    except Exception as e:
        print(f"Error reading file: {e}")

if __name__ == "__main__":
    count_tokens()
```
